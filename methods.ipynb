## Methods 


```python
class Student:
    def __init__(self, name, age, number):
        self.name = name
        self.age = age
        self.number = number

    def display_info(self):
        return f"Name: {self.name}, Age: {self.age}, Number: {self.number}"

    def celebrate_birthday(self):
        self.age += 1
        return f"Happy birthday {self.name}! You are now {self.age} years old."

    def update_number(self, new_number):
        self.number = new_number
        return f"{self.name}'s student number has been updated to {self.number}."


student1 = Student("leo", 18, "n001")
student2 = Student("Brian", 19, "n002")
student3 = Student("fred", 20, "n003")

print(student1.display_info())
print(student2.display_info())
print(student3.display_info())
print(student1.celebrate_birthday())
print(student2.update_number("S099"))
```

##  phone


```python
class phone:
    def __init__(self, brand, model, storage):
        self.brand = brand
        self.model = model
        self.storage = storage
    def display_info(self):
        return f"Brand: {self.brand}, Model: {self.model}, Storage: {self.storage} GB"

    def phone_call(self):
        return f"{self.brand} {self.model} is making a call."
    
phone1 = phone("Apple", "iPhone 14", 128)
phone2 = phone("Samsung", "Galaxy S23", 512)

print(phone1.display_info())
print(phone2.display_info())
print(phone1.phone_call())
print(phone2.phone_call())
```

## OBJECT STATEMENT

- bank account class with ownwer name and starting balance
- owner's name and starting balance 
- add a method to deposit money, another to display account balance 
- create one accout deposit 50000 ugx and display the balance 



```python
class BankAccount:
    def __init__(self, owner_name, account_number, balance=0.0):
        self.owner_name = owner_name
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"Deposited {amount:.2f} UGX. New balance: {self.balance:.2f} UGX"

    def display_balance(self):
        return f"Account Name: {self.owner_name}, Account Number: {self.account_number}, Balance: {self.balance:.2f} UGX"


account1 = BankAccount("leo", "123456789", 1000.00)
print(f"Original balance: {account1.display_balance()}")
print(account1.deposit(50000))
print(account1.deposit(70000))
print(account1.display_balance())
```
