## abstract classes
- these are class that cannot be instianeted on their own   
- they are meant to be subclassed 
- #### benefits 
- prevents instantantiation of the class itself
- requires children to use intenede abstract method


```python
from abc import ABC, abstractmethod

class vehicle(ABC):
    @abstractmethod
    def stop(self):
        pass
    @abstractmethod
    def start(self):
        pass
    
    
class Car(vehicle):
    def stop(self):
        return "Car is stopping"
    
    def start(self):
        return "Car is starting"
    
    
car = Car()


car.start()
car.stop()


```
