# Encapsulation

Encapsulation means keeping data and the methods that work on that data together in one class, while controlling direct access to important values.

## Why it matters

- Protects data from accidental changes
- Keeps the internal structure of a class hidden
- Lets you validate values before saving them
- Makes code easier to maintain and debug

## Example idea

A student class can store a student name and CGPA, but the CGPA should only be allowed inside the valid range of 0.0 to 5.0.
If someone tries to assign 20.0, the class should reject it.

## Access levels in Python

| Level | Example | Meaning | Behavior |
|---|---|---|---|
| Public | `self.name` | Can be used anywhere | Direct access |
| Protected by convention | `self._cgpa` | Internal use, not meant for outside use | Python does not block it, but it signals caution |
| Private by name mangling | `self.__secret` | Stronger internal hiding | Python changes the name to make accidental access harder |
| Controlled access | `@property` and setter | Lets you check values before storing them | Best for validation like CGPA limits |

## Important points to remember

- Use a single underscore for attributes you want to treat as internal.
- Use properties when you want to control reading or writing of a value.
- Encapsulation does not make data impossible to access; it makes access more disciplined.
- Validation should happen inside the class, not from outside the class.

## Memory trick

Think of encapsulation as a **capsule**:
- The outside sees only the label.
- The inside contains the important medicine.
- You do not open it directly unless the class allows it.

A simple memory shortcut is **H-V-C**:
- **H**ide data
- **V**alidate values
- **C**ontrol access

## Quick recap

Encapsulation = hide internal data + protect it with rules + use the class to control changes.
